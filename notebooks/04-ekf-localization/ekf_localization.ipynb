Your exercise for this learning experience will be to write an extended Kalman filter (EKF) for map-based localization.

# Defining the map

We need to input the true locations of the landmarks that we are going to use for localization. They are read from [the map file](../../packages/ekf_localization/map.yaml), which by default contains the landmarks of the Duckiematrix map, so if you are running in the Duckiematrix this is done for you. If you are running on the real robot, you will need to measure the positions of the landmarks in your Duckietown and write them into a map file: either edit `map.yaml`, or add a new file (e.g. `my_town.yaml`) next to it and select it with the `map_name` argument of the [launch file](../../packages/ekf_localization/launch/ekf_localization_node.launch.py), e.g. `ros2 launch ekf_localization ekf_localization_node.launch.py veh:=ROBOT_NAME map_name:=my_town`. Since `dts code workbench` starts the launch file without this argument, change the default value of `map_name` from `map` to `my_town` in the launch file to use your map from the workbench (then rebuild). The same directory also contains [aa3341.yaml](../../packages/ekf_localization/aa3341.yaml), an example of a map measured in a real room (the UdeM AA-3341 lab). The format of the map file is as follows:

```yaml
map:
  "1":
    position: [0.6786, 1.76085]
  "20":
    position: [-0.02925, 1.8252]
  ...
```

The numbers e.g. "1" are the IDs of the [AprilTags](https://april.eecs.umich.edu/software/apriltag). You should also be careful to define origin of the world frame. In the Duckiematrix, the yellow duckie starts at the origin of the world frame pointing in the direction of positive 'y' with positive 'x' to the right.

You may need to [assemble some of your own traffic signs](https://docs.duckietown.com/ente/duckietown-manual/30-duckietown-city/assembly/traffic-signs/duckietown-smart-city-traffic-signs-assembly-instructions.html) to have enough for a good localization performance. 

# Defining the filter parameters

Your EKF requires certain [parameters to be defined](../../packages/ekf_localization/config/ekf_localization_node/default.yaml). These include the initial pose and covariance of the Duckiebot, as well as the process and measurement covariances. Again, if you are running in the Duckiematrix, you can probably leave these alone since we have already specified the initial robot pose to coincide with the location of the Duckiebot in the Duckiematrix. All of them are ROS 2 parameters of the `ekf_localization_node`, so you can also change them while the node is running, from the parameter panel at the end of this notebook or with `ros2 param set` in a terminal of the editor (the `Q`/`R` covariances take effect immediately, the initial pose and covariance at the next `reset`). In the real Duckietown you will need to make sure that you initially place the Duckiebot at roughly the position that is specified in your config. 

**Note**: You should return the Duckiebot to the initial pose every time you test your code. In the Duckiematrix you can do this by pushing the `R` key.


# Implementing your EKF

All the code that you will have to write is in the [ekf.py](../../packages/solution/ekf.py) file (if you are interested, you can also look at the details of how the [ROS node](../../packages/ekf_localization/ekf_localization/ekf_localization_node.py) is implemented). 

The pose of the Duckiebot is stored in `self.q` and is a numpy array with the form
$$
[x, y, \theta].
$$

The state covariance of the Duckiebot is stored in `self.P` and has the structure

$$
P = 
\begin{bmatrix}
P_{xx} &  P_{xy} & P_{x\theta} \\
P_{xy} & P_{yy} & P_{y\theta} \\
P_{x\theta} & P_{y\theta} & P_{\theta\theta}
\end{bmatrix}.
$$

There are two functions to write: `predict` and `update`. 


## The prediction step

The predict function 

```python
def predict(self, dX, dT):
```

takes as input `dX`, which is the linear forward movement of the Duckiebot since the last time the function was called (in meters), and `dT`, which is the amount that the Duckiebot has turned since the last time the function was called (in radians). These values are calculated from the encoder data and act as a proxy for our input (control).

The process model covariance is stored in `self.Q` and has the structure

$$
Q = 
\begin{bmatrix}
Q_{xx} & 0 \\
0 & Q_{\theta\theta}
\end{bmatrix},
$$

where $Q_{xx}$ and $Q_{\theta\theta}$ are covariances specified [in the config file](../../packages/ekf_localization/config/ekf_localization_node/default.yaml).

# Testing and debugging

Once you have completed the implementation you can proceed to test your code using the approach described in the [README](../../README.md): `dts code build -R ROBOT_NAME`, then `dts code workbench [-m] -R ROBOT_NAME`. This runs the `ekf_localization_node` (together with a dead-reckoning reference node, `encoder_pose_node`) on the robot. The node publishes (all names relative to the robot namespace `/ROBOT_NAME/`):

- `ekf_localization_node/pose` (`nav_msgs/Odometry`): your EKF estimate with its covariance;
- `map_markers` (`visualization_msgs/MarkerArray`): the landmarks of the map, green by default and blue while they are being detected;
- `ekf_localization_node/debug/detections/compressed` (`sensor_msgs/CompressedImage`): the camera image with the detected AprilTags superimposed;
- `encoder_pose_node/encoder_pose` (`nav_msgs/Odometry`): the pose obtained from the encoders only (dead reckoning), for comparison.

The cells below connect this notebook to the robot and show these topics live. Set `ROBOT` to the name of your (real or virtual) Duckiebot, run the cells while the workbench is running, then drive the robot around (keyboard of the Duckiematrix, or `dts duckiebot keyboard_control ROBOT_NAME`). If your estimator is performing well, you should see that the ground truth state remains inside the covariance ellipse of the estimate that you are producing as you pilot your vehicle around.
````python
from lx_ros2_tools.session import RosSession
from lx_ros2_tools.panel import ActionButtons, ImageView, ParamPanel, TrajectoryPlot

ROBOT = "ROBOT_NAME"   # <-- the name of your (real or virtual) Duckiebot
ros = RosSession(ROBOT)  # connects to the robot's zenoh router and spins in the background
ros
````

````python
# the EKF estimate (x, y and heading), one point per published pose
traj = TrajectoryPlot(ros, "ekf_localization_node/pose", title="EKF estimate")
````

````python
# landmarks of the map and the 3-sigma covariance ellipse of the estimate, drawn on the same axes
import math
import numpy as np
from matplotlib.patches import Ellipse
from nav_msgs.msg import Odometry
from visualization_msgs.msg import MarkerArray

(landmarks_all,) = traj.ax.plot([], [], "s", color="green", ms=7, label="landmarks")
(landmarks_seen,) = traj.ax.plot([], [], "s", color="blue", ms=7, label="detected now")
ellipse = Ellipse((0.0, 0.0), 0.0, 0.0, fill=False, color="purple", lw=1.5, label="3-sigma covariance")
traj.ax.add_patch(ellipse)
traj.ax.legend(loc="upper left", fontsize="small")


def on_markers(msg: MarkerArray):
    pts = [(m.pose.position.x, m.pose.position.y, m.color.b > 0.5) for m in msg.markers]
    landmarks_all.set_data([p[0] for p in pts], [p[1] for p in pts])
    landmarks_seen.set_data([p[0] for p in pts if p[2]], [p[1] for p in pts if p[2]])


def on_pose(msg: Odometry):
    P = np.array(msg.pose.covariance).reshape(6, 6)[:2, :2]
    vals, vecs = np.linalg.eigh(P)
    vals = np.clip(vals, 0.0, None)
    ellipse.center = (msg.pose.pose.position.x, msg.pose.pose.position.y)
    ellipse.width, ellipse.height = 2 * 3 * math.sqrt(vals[1]), 2 * 3 * math.sqrt(vals[0])
    ellipse.angle = math.degrees(math.atan2(vecs[1, 1], vecs[0, 1]))


sub_markers = ros.subscribe("map_markers", MarkerArray, on_markers)
sub_pose = ros.subscribe("ekf_localization_node/pose", Odometry, on_pose)
# (the plot above refreshes every time a new pose is published: after every prediction while the robot moves,
# and at least twice per second (update step) when it stands still)
````

## Debugging

If you have made an error in your implementation, it is likely that your estimate will diverge very quickly. Debugging this can actually be quite challenging since the divergence happens quickly and it is difficult to determine the root cause. One thing that can help is to isolate the prediction and the update steps. Specifically, you can test each step independently by setting `no_update` or `no_predict` to `true` (in the parameter panel above, with `ros2 param set /ROBOT_NAME/ekf_localization_node no_update true` in a terminal of the editor, or in the [config file](../../packages/ekf_localization/config/ekf_localization_node/default.yaml) before rebuilding). Be somewhat careful with the `no_predict: true` case however, since you are getting no information about how the robot is moving and relying entirely on the landmark measurements. If you have some period of time without landmark measurements your estimate will not be updated. Therefore a recommended method to proceed if you find that your estimator is not performing well is to begin with `no_update: true` until you are confident that your predict step is correct and then set `no_update: false` (leaving `no_predict` as `false` the entire time).

Other useful tools:

- compare your estimate with the dead-reckoning pose (`encoder_pose_node/encoder_pose`) and, in the Duckiematrix, with the ground truth (`duckiematrix_interface_node/state`): the cell below plots them;
- a terminal inside the editor is connected to the robot: `ros2 topic list`, `ros2 topic echo /ROBOT_NAME/ekf_localization_node/pose`, `ros2 topic echo /ROBOT_NAME/map_markers --once`, `ros2 param list /ROBOT_NAME/ekf_localization_node`, `ros2 service call /ROBOT_NAME/ekf_localization_node/reset std_srvs/srv/Trigger`;
- the log of the node is printed by `dts code workbench`.

If you still have difficulty understanding what is going wrong, you may find it helpful to print out the intermediate results from the steps above and verify that they are correct one by one.
